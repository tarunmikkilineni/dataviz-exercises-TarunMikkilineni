```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv(
    '/kaggle/input/datasets/ajaypalsinghlo/world-happiness-report-2023/WHR2023.csv'
)

print(df.columns.tolist())
df.head()
```

```python
df.head()
```

```python
# Sort countries by happiness score
df_sorted = df.sort_values(
    by='Ladder score',
    ascending=False
)

# Select the 10 happiest and 10 least happy countries
happiest = df_sorted.head(10)
least_happy = df_sorted.tail(10)

# Create charts
fig, axes = plt.subplots(1, 2, figsize=(16, 6))

axes[0].barh(
    happiest['Country name'],
    happiest['Ladder score']
)
axes[0].set_title('Top 10 Happiest Countries')
axes[0].set_xlabel('Happiness Score')
axes[0].invert_yaxis()

axes[1].barh(
    least_happy['Country name'],
    least_happy['Ladder score']
)
axes[1].set_title('10 Least Happy Countries')
axes[1].set_xlabel('Happiness Score')

plt.tight_layout()
plt.show()
```
